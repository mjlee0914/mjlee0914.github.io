### 2. Type Conversions
#### Explicit type conversion
- `int()`
- `float()`
- `str()`
- `bool()`
- `ord()`

#### Implicit type conversion
- 더 작은/낮은 범위의 타입을 **더 넓은 타입**으로 자동 변환 ex) int + float = float
- type upcasting: float은 데이터 표현 범위가 int보다 **더 넓음** -> 항상 float type으로 upcast

```python
# type upcasting example
x = 10
y = 3.14
print(type(x+y)) # <class 'float'>
```

### 3. ValueError
- type conversion을 할 때 혀여태에 맞지 않는 부적절한 데이터를 강제로 바꾸면 ValueError 발생
- str.isdigit() : True / False

```python
val = "20 year-old"
int(val) #ValueError
```

```python
age = input("Enter your age: ")
if age.isdigit():
    age = int(age)
    print(f"You are {age} years old.")
else:
    print("Not a valid answer.")
```

### 3. Comparison Operators
- `>` : greater than
- `<` : less than
- `>=` : greater than or equal to
- `<=` : less than or equal to
- `==` : equality operator
- `!=` : not equal operator

**Equality operator vs identity operator**
- `==` : same value?
- `is`(identity operator) : same object? same address?

```python
a = [1, 2, 3]
b = [1, 2, 3]
print (a == b) 
print (a is b)
# same value, different address (object)
```

### 4. Logical Operators
- `and` : both should be `True` = `True`
- `or` : one is `True` = `True`
- `not` : reverse bool

```python
is_a = True
is_b = False
print(not is_a)
print(is_b)
```

### 5. Short-Circuit Evaluation
- `and` 
    - 왼쪽이 `True` → 오른쪽 평가
    - 왼쪽이 `False` → 오른쪽 평가하지 않음
    - `False`를 만나면 결과가 결정됨

- `or`
    - 왼쪽이 `True` → 오른쪽 평가하지 않음
    - 왼쪽이 `False` → 오른쪽 평가
    - `True`를 만나면 결과가 결정됨

```python
# and
print(True and 10 > 5)
print(False and 1 > 0)
# or
print(True or 10>5)
print(False or 10>5)
print(False or 10>11)
```
