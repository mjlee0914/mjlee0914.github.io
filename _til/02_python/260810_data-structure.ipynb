*** Nested Data Structure *** 
- list, tuple, dictionary 모두 다른 객체를 요소로 가질 수 있음

*** index ***
- [0, 1, 2, 3, ...]
- starts with 0

*** Shallow Copy ***


```python
x = [0, 1, 2, 'apple', True, False, [3, 4, 5], (6, 7, 8)]
x2 = x[:] # x2 = x.copy()와 같음

print(x2[-1])
print("-2번째 index의 리스트의 index 0 접근: ", x2[-2][0])

x3 = x.copy()
print(id(x), id(x2), id(x3)) #print한 모든 변수의 id 값이 다름 --> copy()는 새로운 객체 생성

# Shallow Copy
print(id(x) == id(x2)) # x2는 x를 copy()하여 참조하였으므로, 서로 id값 자체는 다름. copy()는 바깥 쪽 리스트만 새로 만듦
print(id(x[-2][0]) == id(x2[-2][0])) #그러나 x와 x2가 내부적으로 참조하는 mutable 객체 안 값의 주소는 같으므로 True

x2[-2].append(6)
print(x2)
print(x) # mutable한 내부 list 객체를 공유하는 상태에서 새로운 값을 수정(appened 등)할 경우 (내부적으로 참조하는 mutable 값), 공유하는 다른 리스트의 객체도 똑같이 바뀜
# str, int, tuple같은 immutable 객체의 경우, x2가 수정되었더라도 원본 x에는 영향 없음


x2[-2] = [1, 2] # 다음의 경우 공유된 내부 객체를 수정하는 것이 아닌 참조 자체를 바꾸는 것이므로 (다른 객체를 만드는 것이므로) x에 영향 X
print(x2)
print(x)

```

## 2. List
- mutable
- [val, val2, val3, ...]

Create
- `.append(val)`: 끝에 추가
- `.insert(추가할 위치, 추가할 요소)`
- `extend(collection - list/tuple)`: 원본 리스트 끝에 한꺼번에 여러개 추가
  - `[원본 리스트] + [val1, val2, ...]`: 끝에 여러개 추가, 새로운 리스트 생성
  - `.extend(val) = 'v', 'a', 'l', 'u', 'e'
  - `.extend([val]) = 'value' 

Read
- index number `list(0)`

Update
- `list(0) = "changing val"`

Delete
- `remove()`: 첫 번째 발견한 특정 요소 제거
    - `list.remove(val)` : remove by value (list method)
    - `del list[number]` : delete by index (python statement)
- `clear()`: clear every value

Etc.
- `sort()`: 오름차순 정렬
- `sort(reverse=True)`: 내림차순 정렬
- `reverse()`: 순서 반전
- `count()`: 특정 요소 등장 횟수


## 3. Tuple
- comma!
- immutable
- (val, val2, val3, ...)
- read-only
- tuple[num]: index access
- count(val): val 등장 횟수
- index(val): val의 index
- `(val1,)`: 값이 하나인 튜플은 반드시 `,` 추가
- 소괄호 생략 가능
- 빈자리 = underbar
- `*변수명` = 나머지 underbar 여러개 대신 사용 가능

## 4. Indexing
- index[0, 1, 2, 3] = index[-4, -3, -2, -1]

## 5. Slicing
- 결과 -> new list / new tuple
- `[start:]` -> start ~ end
- `[:end]` -> start ~ 종료앞
- `[start:end]` -> 시작 인덱스 이상, 종료 인덱스 미만
- `[start:end:step]` -> 시작지점에서 종료 지점 미만까지 증감 단위로 추출 
- `[::-1]`: 전체 리스트 reverse 접근

## 6. sequence operators
- `+`: 새로운 list/tuple 추가
- `*number`: 숫자만큼 반복해서 새로 생성
- `val in sequence`: is val included? (Bool)
  - `val not in sequence`
- `len()`: number of val, size of list/tuple

## 7. packing and unpacking
- explicit type change
  - `list(...)`: from tuple/set to new list
  - `tuple(...)`: from list/set to new tuple

*** list unpacking ***
- assign the value to several variables
- `*val`: other values, combined to list 

```python
fruits = ['apple', 'banana', 'kiwi']
apple, *restFruits = fruits
print(apple, restFruits)
```

```python
packed_data = 1, 2, 3 #packing several values into one tuple
print(packed_data)

a, b , c = packed_data # unpacking the tuple into several variables
print(a, b, c)

first, *others = [1, 2, 3, 4] #unpacking using *
print(first, others)

# one variable = one/the first value
# *variable = remaining values as a list
```

# 8. Dictionary
- {key:value, key:value}
- `key`: unique, List unavailable, tuple available(immutable)
- `value`: duplicatable


Create
- `dict['key'] = val`

Read
- `dict['key']` : if not, `KeyError`
    - `.get('key')`: if not, `None`
    - `.get('key', 'default')`: if not, default
- `.keys()`: every key, read-only
- `.values()`: every value, read-only
- `.items()`: (key, value) view

Update
- `.update({key: val, key: val, ...})`

Delete
- `del items['key']`: delete key-value. statement-> cannot be assigned to a variable
- `.pop()`: delete key-value pair, return the deleted value that can assigned to a variable
- `.pop('key', default value)
- `.clear()`: clear everything

### hash vs. id
- hash('key'): hash value of the 'key' -> decide key saving/search location from hash table
  - hash collision: differnet obj, same hash value
- id('key'): identity of the 'key', to check if the two variables refer the same obj



# 9. Set
- unique
- no index, no location
- {val1, val2, val3, ...}
- `set()`: from tuple/list to set

## set operation
- `a | b`: `a.union(b)` exisit both in a or b
- `a & b`: `a.intersection(b)` exist both in a and b
- `a - b`: `a.difference(b)` exist in a but not in b
- `a ^ b`: `a.symmetric_difference(b)` exist in either a or b
