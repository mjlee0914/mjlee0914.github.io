### 2.2 매개변수 Parameter and 인수 Argument
- Parameter: 함수의 '선언부' 공간에 위치
- Argument: 실제 함수 작동 시 해당 통로를 통해 전달하는 '실제 데이터', 함수 '호출부' 공간에서 주입

````python
def taste(food): # food = parameter
    print(f"{food}는(은) 맛있다")

taste("떡볶이") # arguments
taste("사과")
````

### 2.3 위치 인수 Positional Arguments and 키워드 인수 Keyword Arguements
위치 인수
- parameter 순서대로 인수를 차례대로 주입
- 순서 바뀌면 오작동 유발 가능 주의

키워드 인수
- `매개변수명=값` 형태
- 순서 상관 X
- 위치 인수와 작성시 반드시 위치 인수 먼저!

````python
def calc(num1, num2):
    return num1+num2

print(calc(1, 2)) # 위치 인수   
print(calc(num2=20, num1=10)) # 키워드 인수
````

### 2.4 Default Parameter 기본값 매개변수
- 함수 정의 시 매개변수에 특정 값을 미리 대입 
- 특정 값이 매겨진 매개변수는 함수 호출 시 해당 인수 생략 가능
- 기본값이 없는 매개변수보다 뒤에 위치해야 함 (interpreter가 인수를 앞에서부터 매핑함)

````python
def introduce(name, age=20):
    print(f"이름은 {name}이고, 나이는 {age}세입니다.")

introduce("철수")       # 나이 인수를 생략하면 기본값 20이 자동 적용
introduce("지영", 25)   # 인수를 전달하면 25로 갱신되어 작동
````

### 2.5 가변 인수 *args, 키워드 가변 인수 **kwargs
- `*args`
  - 입력될 인수가 몇 개인지 아직 정해지지 않았을 때 매개변수에 적용
  - 출력시 하나의 **튜플 구조**로 묶여 차례대로 순회가 가능

- -`*kwargs`
  - 인수 이름을 지정하여 '키-값' 형태로 전달
  - 입력된 모든 쌍 데이터를 하나의 **딕셔너리 구조**로 변환하여 처리

````python
def print_user_info(name, age, *args, **kwargs):
    print("name:", name)
    print("age:",  age)
    print(args)
    print(kwargs)

print_user_info("김철수", 20, 'A', 20, mobie="010-0000-0000") # 'A', 20은 Tuple로 *args에 들어감
````

### 2.6 return
- 함수 내부에서 연산된 최종 결과값을 내보내는 통로 역할
- interpreter가 return 만나면 즉시 함수 종료&탈출
- **복수 데이터 반환** 가능: **튜플** 구조로 묶임, unpacking 가능

````python
def calc(num1, num2):
    return num1 + num2, num1 * num2, num1 - num2, num1 / num2 # return 값 여러개면 튜플 하나로 반환

print(calc(2,2))
x = a, b, c, d = calc(2,2) # unpacking
print(x)
````

## 3. [심화] 파이썬 함수의 메모리 생애주기 (Heap vs Call Stack)
***함수의 정의(코드)는 힙(Heap)에 상주하고, 함수의 실행(상태)은 콜 스택(Call Stack)에 임시 생성되었다가 사라진다***

| 구분 | 힙(Heap) 영역 | 콜 스택(Call Stack) 영역 |
|---|---|---|
| **저장 대상** | 함수 정의 객체 (`PyCodeObject`, `PyFunctionObject`) | 함수 실행 프레임 객체 (`PyFrameObject`) |
| **생성 시점** | 모듈 로드 및 `def` 문 해석 시점 | 함수 호출 시점 (`func()`) |
| **소멸 시점** | 프로그램 종료 또는 가비지 컬렉션(GC) 시점 | 함수 실행 완료 시 (`return`) 즉시 POP |
| **특징** | 함수를 여러 번 호출해도 단 1개의 원본 객체를 재사용 | 호출할 때마다 새로운 독립 프레임 생성 (재귀 호출 포함) |

#### 구조
- `정적 메모리`: 한번 기록되면 지워지지 않음
  - 코드 영역
  - 전역변수
  
- `동적 메모리`: 실행 중 객체 생성/소멸 반복, 안쓰면 제거됨
  - `Heap memory`: 객체가 주로 존재, PyFunctionObject-PyCodeObject
    - a variable created -> its object/value is allocated in Heap memory ex) large data structures(lists, dictionaries, objects)
    - only the reference of the object is stored in stack memory
    - objects in Heap memory can be shared among multiple functions even after a function finished executing
    - Garbage Collection -> clean up unused object from Heap memeory (regarding reference counter) after the stack memory frees the function frame

- `call stack`: 함수 호출 시 실행, 상태 관리 영역
  - whenver a function is called, Python creates a stack frame(PyFrameObject) for the function and add it to the call stack.
  - the local variables(references to objects) declared inside the function (local variable) are stored inside the function's stack frame
  - after finishing executing, stack memory is freed automatically

#### 단계별 메모리 적재 프로세스
1. 정의 단계 def
   
     파이썬 인터프리터가 `def` 문을 실행하면 함수의 코드 정보를 담은 **`PyCodeObject`**와 이를 감싸는 **`PyFunctionObject`**가 생성되어 힙(Heap)에 저장된다.

    - `PyCodeObject` → 함수의 실행 코드 정보
    - `PyFunctionObject` → 함수 객체
    - 함수가 호출되지 않아도 **함수 정의 자체는 먼저 생성**됨

2. 호출 단계 function call

     함수가 호출되면 새로운 **`PyFrameObject`**가 생성되어 콜 스택(Call Stack)에 PUSH된다.

    - `func()` 호출 → 새로운 실행 프레임 생성
    - `PyFrameObject` → 해당 호출을 위한 **독립적인 실행 환경**
    - 매개변수와 지역 변수 등은 프레임의 실행 상태에 저장됨
    - 같은 함수를 여러 번 호출하면 **호출마다 새로운 프레임**이 생성됨
    - 재귀 호출에서도 호출 횟수만큼 독립적인 프레임이 생성됨
  
3. 종료 단계 return

     함수 실행이 끝나면 해당 실행 프레임이 콜 스택에서 POP되고, 함수 호출에 사용되던 실행 상태가 정리된다.

    - `return` → 현재 `PyFrameObject`가 콜 스택에서 제거됨
    - 지역 변수도 해당 프레임에 대한 참조가 사라지면서 정리될 수 있음
    - 힙에 존재하는 **`PyCodeObject`와 `PyFunctionObject`는 그대로 유지**
    - 따라서 같은 함수를 다시 호출하면 **기존 함수 객체를 재사용하면서 새로운 프레임을 생성**

### 핵심 흐름

`def` → **함수 객체 생성 (Heap)**  
↓  
`func()` → **실행 프레임 생성 및 PUSH (Call Stack)**  
↓  
`return` → **실행 프레임 POP**  
↓  
다시 `func()` → **새로운 실행 프레임 생성**

> **함수 자체는 Heap에 존재하며 재사용되고, 함수 호출마다 새로운 실행 프레임이 Call Stack에 생성된다.**

#### 전역변수: 함수 외부에 정의된 변수 
#### 지역변수: 함수 내부에 정의된 변수. 매개변수도 지역변수에 해당

````python
def add(num1, num2): #매개변수==지역변수. 
    num3 = 30 #지역변수

    result = num1 + num2 + num3 # result = local variable

    return result 

result = add(10,20) # result = global variable

print(result)


#1. add라는 이름이 Heap memory에 있는 Function Object를 참조함.
#2. Function Object는 Code Object 및 함수 관련 정보를 가짐
#3. id(add)로 Function Object의 identity를 확인할 수 있음.
# num1, num2는 함수 호출시 Frame에서 지역 이름으로 관리됨.
# 10과 20은 Heah에 존재하는 int 객체



#      함수 내부                         함수 외부

#    add Frame                           Global
# ┌──────────────┐                  ┌──────────────┐
# │ num1 → 10    │                  │ result → 60  │
# │ num2 → 20    │                  └──────┬───────┘
# │ num3 → 30    │                         │
# │ result → 60  │                         ↓
# └──────────────┘                       Heap
#                                        60 객체
````

````python
# int id
x = 10
y = x

x += 1

if id(x) != id(y):
    print("x and y do not refer to the same object")

# list id
a = [1, 2, 3]
b = a
print(f"a ID: {id(a)}, b ID: {id(b)}") # Same ID → both point to same list 
b.append(4) 
print(f"a: {a} b: {b}")
````

````python
def add(a, b):
    result = a + b
    return result

def calc():
    x = 10
    y = 20
    return add(x, y)


# mental model of memory process

# FUNCTION DEFINITION
#            ↓
#    Function object exists in heap/object memory
#            ↓
#    CALL calc()
#            ↓
#    ┌───────────────────────┐
#    │ calc() stack frame    │
#    │ x → object            │
#    │ y → object            │
#    └───────────────────────┘
#            ↓ calls
#    ┌───────────────────────┐
#    │ add() stack frame     │
#    │ a → object            │
#    │ b → object            │
#    │ result → object       │
#    └───────────────────────┘
#            ↓
#    add() finishes
#            ↓
#    add() frame removed
#            ↓
#    calc() continues
#            ↓
#    calc() finishes
#            ↓
#    calc() frame removed
###
````

````python
# 문제 3-1: 무제한 과목 점수 합계 및 평균 산출
def get_total_and_average(*arg):
    if len(arg) == 0:
        return 0
    else: 
        total = sum(arg)
        avg = total / len(arg)
        return total, avg
       

print(get_total_and_average(90, 85, 95)) 
print(get_total_and_average(70, 80, 90, 100, 65))
````
