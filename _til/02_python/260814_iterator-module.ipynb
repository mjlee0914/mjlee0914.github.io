## 2. Key Concepts

## 2.1 iterable, iterator
iterator
- list, str, dict, ... iterable 객체: 내부적으로 `\_\_iter\_\` 메서드 가짐
- 내부 요소를 하나씩 꺼내서 순회 가능
- `\_\_iter\_\` = `iter()`
- `\_\_next\_\` = `next()`: 한 번에 하나씩 데이터를 메모리에 불러와 처리, 더이상 불러올 데이터 없으면 'Stopiteration' 예외 발생

```python
# 내장 리스트 객체(Iterable) 생성
fruits = ["사과", "바나나", "체리"]

# iter() 함수를 사용하여 이터레이터 객체 추출하기
fruits_iterator = iter(fruits)
print(f"이터레이터 객체 확인: {fruits_iterator}")

# next()를 호출하며 원소를 순차 수동 추출하기
print(next(fruits_iterator))  # 사과
print(next(fruits_iterator))  # 바나나
print(next(fruits_iterator))  # 체리

# 더 이상 꺼낼 원소가 없을 때 예외 발생 확인
# print(next(fruits_iterator))  # StopIteration 예외 발생
```

## 2.2. generator and yield keyword

generator function
- Lazy Evaluation: 호출될 때마다 값을 하나씩 생성해서 내보냄, 필요할 때만 수행
- 메모리 절약

`yield` keyword
- return과 차이: return은 최종 결과값 반환 + 함수 공간 전체 메모리에서 소멸, yield는 결과값 함수 밖으로 반환 + 실행 상태를 메모리에 얼려둔 채로 잠시 대기
- 호출 뒤 다시 외부에서 데이터 요청 by `next()` -> `yield`는 얼어있던 상태를 깨움 -> yield 다음 줄부터 다시 실행


### list comprehension vs. generator expression
list comprehension: 결과물을 모두 메모리에 일괄 적재

- `[표현식 for in ~]`

generator expression : lazy evaluation -> 메모리 절약

- `(표현식 for in ~)`

  

## 2.3. module and package
- module : 하나의 파이썬 파일, 필요할 때 파일 안에 있는 변수, 함수, 클래스 등을 호출하여 재사용 가능 or 수정 가능
- package : 모듈을 폴더 형태로 계층화하여 묶은 구조
- `.`: 현재 내가 속한 폴더 위치
- `..`: ..상위 폴더 위치


모듈 로드 방법
- `import 모듈`: 모듈.함수()로 모듈에 있는 함수 동작    
- `from 모듈/패키지 import 함수/변수/클래스/.../모듈`: 함수()로 함수 동작
- `from 모듈 import *` : 모듈 전체 
- `import 모듈 as 별명`


#### \_\_init\_\_.py
- 현재 디렉토리가 패키지임을 알려주는 파일(디렉토리 안에 생성)
- 모듈 불러올 때 최초 한 번만 실행
- 내용: 패키지 내 사용하는 공통 함수, 변수, 클래스 등 정의
- from 패키지 import * 했을 때 __all___ = [....] 하면 무엇을 가져올지 [....] 안에 지정 ex) \_\_all\_\_ = ["math", "string"]

#### \_\_pycache\_\_/
- import한 모듈의 바이트코드 캐시 저장소
- import 다시 할 때 조건 맞으면 해당 캐시 재사용 -> 불필요한 재컴파일 줄임

```python
# import 키워드를 사용한 전체 모듈 로드 방식
import math
print(math.sqrt(16))  # 모듈명을 명시하여 식별

# from 키워드를 사용한 특정 기능 로드 방식
from math import sqrt
print(sqrt(25))       # 모듈명 없이 함수 이름만으로 곧바로 실행

# as 키워드를 사용한 별칭 부여 방식
import math as m
print(m.sqrt(36))     # 설정한 별칭 m을 사용하여 연산 수행
```

## 2.4. synchronous vs. asynchronous
동기 synchronous

- 순서대로 실행하고 결과도 순서대로 도출 (이전 작업 종료 -> 다음 작업 시작)
- a 작업 시작 -> a 작업 종료 -> b 작업 시작 -> b 작업 종료 -> c 작업 시작 ...

비동기 asynchronous

- 순서대로 작업을 시작하지만 결과는 먼저 처리된 순서대로 나오는 방식
- 작업 완료시 '이벤트 발생'
