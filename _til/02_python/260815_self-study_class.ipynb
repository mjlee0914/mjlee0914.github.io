```python
class Book:
    def __init__(self, title, author, isbn):
        self.title = title
        self.author = author
        self.isbn = isbn

    def show_title(self):
        print(f"Title of the book is {self.title}")

book_1 = Book("Pachinko", "이민진", "1234")
book_1.show_title()
```

```python
class human:
    def __init__(self, name, nationality, language): # 객체의 초기 상태 설정, 인스턴스가 가져야 할 고유한 데이터를 선언
        self.name = name
        self.nationality = nationality
        self.language = language

    def __str__(self): # print(객체)
        return f"name = {self.name} | natioinality = {self.nationality} | language = {self.language}"
    
    def __repr__(self): # 객체 이름 or debugger 창에서 변수 볼 때 데이터 상태 확인 용도
        return f"human(name={self.name}, nationality={self.nationality}, language= {self.language})"

    def introduce(self): # 인스턴스 메서드 (self 매개변수로 받음. 객체의 데이터를 다룸)
        print(f"Hello! My name is {self.name} and I'm from {self.nationality}. I speak {self.language}. ")
    
    

me = human("MJ", "South Korea", "Korean and English") # 객체 'me' as instance of the class 'human' 

me

print(me)
print(me.introduce())



```

```python
class Animal:
    def breathe(self):
        print("숨을 쉽니다.")

# Animal 클래스의 기능을 그대로 물려받습니다.
class Dog(Animal):
    def bark(self):
        print("멍멍 짖습니다.")

baduk = Dog()
baduk.breathe()  # 부모에게 물려받은 기능을 그대로 사용합니다.
baduk.bark()    
```
