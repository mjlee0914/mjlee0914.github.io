## 2. Key Concepts


**Object-oriented Programming**
- 작은 문제를 해결할 수 있는 기능(객체)들이 유기적으로 모여 최종 문제해결을 함.
- 함수 중심이 아닌 객체들의 협력으로 프로그램을 만드는 프로그래밍 방식
- 관련된 기능을 하나의 객체로 묶어 관리 -> 코드 재사용, 유지보수 쉬움
- 4개 특징: encapsulation, inheritance, polymorphism, abstarction

**object**: class/instance/method/variables
- 속성(데이터)와 행동(기능)이 있음


**class**
- blueprint for making object(s)
- 객체가 <어떤 속성>을 가지고 <어떤 기능>을 하는지 정하는 틀

**instance**
- class를 바탕으로 만들어진 객체


**\_\_init\_\_**
- 초기화 함수, 인스턴스가 만들어지는 순간 초기 상태 설정 = 데이터(속성) 설정
- 인스턴스 변수에 최초로 값을 할당


**self**
- 클래스 내부에서 매서드의 첫번째 매개변수
- 객체를 가르키는 참조변수


**속성 attribute**
- 클래스 내부
- 객체의 상태를 나타내는 변수

**인스턴스 매서드 instance method**
- self를 매개변수로 받음
- 객체의 데이터를 다루는 함수


**매직 매서드**
- `\_\_method\_\_`
- \_\_init\_\_
- \_\_str\_\_ -> print() 예쁘게 출력 
- \_\_repr\_\_ 
    - print(객체 참조변수)시 자동호출
    - str과 출력시 str이 우선순위 더 높음
    - 로깅 목적
    -`클래스명(매개변수=값)` 형식, `eval()`
    - `repr(객체 참조변수)`로 호출
- \_\_dir\_\_: 객체가 사용가능한 속성, ,메서드 목록 확인, 알파벳순 정렬

```python
class Pokemon:
    def __init__(self, name, pokemon_type):
        self.name = name
        self.pokemon_type = pokemon_type 
        self.level = 1              # class에서 정하는 공통 DEFAULT 값 = 항상 만드는 포켓몬은 level, hp, attack count가 정해진 값에서 시작
        self.hp = 100 
        self.attack_count = 0

    def __str__(self): # print()시 출력
        return f"{self.name} is {self.pokemon_type} type pokemon!"


    def __repr__(self): # 객체 단독 조회/debugger시 출력 | 리스트 안 객체 기본 표현 바꿔줌
        return f"pokemon: {self.name} | type: {self.pokemon_type} | level: {self.level} | hp: {self.hp} | attack count: {self.attack_count}"
    
    def __dir__(self): 
        return [self.name, self.pokemon_type]

    def attack(self):
        self.attack_count += 1

pikachu = Pokemon("pickachu", "electric")
print(pikachu)
print(dir(pikachu)) # dir() -> 알파벳순 정렬
print(repr(pikachu))
```

**상속**
- 부모 클래스를 재사용 -> 개발시간 단축, 코드 유지보수 쉬움

- method overriding : 부모 클래스의 매서드 -> 자식 클래스에서 새롭게 재정의, 자식 클래스 우선

- `super()`
    - 자식 클래스에서 부모 클래스 메서드 or 초기화 함수 호출시 사용
    - 자식 클래스만의 속성 새로 추가 + 부모 초기화 로직 재사용


**캡슐화/은닉**
- 객체 내부 핵심 데이터 보호
- 캡슐화: 서로 관련 있는 데이터/기능을 하나의 클래스로 모음
- 은닉: 외부에서 내부 데이터 조회/접근/수정 차단

방법
- 언더바 2개 사용: name mangling -> 외부에서 원래 이름으로 직접 접근/실수 접근 불가. private variable
- getter/setter 메서드: 객체 내부 private variable 안전한 조회/수정
    - get_속성(self) return self.__속성
    - set_속성(self, 속성) + if 조건 추가

**getter/settter**
통제
- getter: 값을 안전하게 꺼내서 전달
- setter: 올바른 값인지 검증 후 변경

```python
# chap8-1 문제 2
class SmartLamp:
    def __init__(self, model):
        self.model = model
        self.is_on = False

    def on(self):
        self.is_on = True
        print(f"{self.model} 조명이 켜졌습니다.")

    def off(self):
        self.is_on = False
        print(f"{self.model} 조명이 꺼졌습니다.")


smart_lamp = SmartLamp("M05")
smart_lamp.on()
smart_lamp.off()
```

```python
# chap8-1 문제 3
class Hero:
    def __init__(self, name):
        self.name = name
        self.hp = 100

    def take_damage(self, damage):
        self.hp -= damage
        print(self.hp)

        if self.hp <= 0:
            print(f"{self.name} 캐릭터가 쓰러졌습니다.")

cha = Hero("Spiderman")
cha.take_damage(60)
cha.take_damage(20)
cha.take_damage(10)
cha.take_damage(10)
```

```python
# chap8-2 문제 1
class Phone:
    def call(self):
        print("통화를 시작합니다")

class SmartPhone(Phone):
    def call(self):
        print("영상 통화를 시작합니다")

    def search_web(self):
        print("인터넷 검색을 시작합니다")


iPhone = SmartPhone()
iPhone.call()
iPhone.search_web()
```

```python
# chap8-2 문제 2
class DoorLock:
    def __init__(self, password):
        self.__password = password

    def __str__(self):
        return f"현재 비밀번호: {self.__password}"

    def get_password(self):
        return self.__password

    def set_password(self, password):
        if not password > 4:
            print("경고: 비밀번호는 4자리 이상이어야 합니다")

        else: 
            self.__password = password
            print("정상 변경되었습니다")

new = DoorLock(1222)
new.__password = 1234
print(new)

new.get_password()
new.set_password(1)
new.set_password(123456)
print(new)
```

```python
# chap8-2 문제 3
class User:
    def __init__(self, name):
        self.name = name


class VIPUser(User):
    def __init__(self, name, point):
        super().__init__(name)
        self.point = point

    def __str__(self):
        return f"VIP user {self.name} has {self.point} points"


vip = VIPUser("Lee", 50)
print(vip.name)
print(vip.point)
print(vip)
```

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance
    def __str__(self):
        return f"{self.owner}님의 계좌 (현재 잔액: {self.balance}원)"


class MinusAccount(BankAccount):
    def __init__(self, owner, balance, limit):
       
        super().__init__(owner, balance)
        self.limit = limit  

  
    def withdraw(self, amount):
        
        if (self.balance + self.limit) >= amount:
            self.balance -= amount
            print(f"[{self.owner} 마이너스 계좌] {amount}원 출금 완료 / 현재 잔액: {self.balance}원")
        else:
            print(f"[{self.owner} 마이너스 계좌] 출금 실패: 마이너스 한도({self.limit}원)를 초과했습니다.")
        return self.balance
```

```python
class ProtectedAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance  

    def get_balance(self): 
        return self.__balance

    def set_balance(self, amount):  
        if amount >= 0:
            self.__balance = amount
            print(f"[시스템] 잔액이 {amount}원으로 정상 변경되었습니다.")
        else:
            print("[시스템 경고] 잘못된 수치 입력: 잔액은 음수가 될 수 없습니다.")
```

```python

acc = ProtectedAccount("이몽룡", 3000)


acc.__balance = -5000 
print(f"직접 수정 시도 후 조회 결과: {acc.get_balance()}원")  


acc.set_balance(-1000)  
acc.set_balance(5000)   
print(f"최종 안전 조회 결과: {acc.get_balance()}원") 
```
