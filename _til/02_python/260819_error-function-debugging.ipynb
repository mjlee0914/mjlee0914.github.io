#### @classmethod, @staticmethod
**instance method**
- 클래스 내부에서 정의
- 클래스를 통해 생성된 객체(인스턴스)에서 호출
- 첫 번째 매개변수 self, 호출 시 인스턴스를 전달

**class method**
- `@classmethod` decorator 이용해서 정의
- 클래스 변수 조작 가능
- 새로운 인스턴스 생성 가능
- 첫 번째 인자로 cls 받음 -> 클래스 자체 전달 == 클래스 전체에 관련된 작업

**static method**
- `@staticmethod` decorator 이용하여 정의
- self, cls 받지 않음
- 클래스 내부 데이터 접근 불가 (클래스/인스턴스 관련 작업 안함)
- 인스턴스 속성, 인스턴스 메서드 필요 없을 때 사용
- 메서드 실행이 외부 상태 영향 미치지 않는 순수 함수 만들 때 사용

```python
class Pokemon:
    def __init__(self, name, type): # 인스턴스 생성 시 호출. name, type 초기화
        self.name = name
        self.type = type

    def hungry(self): # instance method
        print(f"{self.name}이(가) 배가 고픕니다")

    def info(self): # instance method 
        print(f"포켓몬: {self.name}")
        print(f"타입: {self.type}")

pikachu = Pokemon("Pikachu", "Electric") # instance from class 'Pokemon', having 2 attributes 'name', 'type'
```

```python
class Student:
    school = "Old school" # 클래스 변수

    @classmethod
    def change_school(cls, new_school):
        cls.school = new_school

print(Student.school)
Student.change_school("New school")
print(Student.school)
```

```python
class Product:
    tax_rate = 0.10

    def __init__(self, name, price):
        self.name = name
        self.price = price

    # 1. Instance method 
    # product instance 
    def price_with_tax(self):
        return self.price * (1 + Product.tax_rate)

    # 2. Class method
    @classmethod
    def change_tax_rate(cls, new_rate):
        cls.tax_rate = new_rate

    # 3. Static method
    @staticmethod
    def is_valid_price(price):
        return price >= 0



apple = Product("Apple", 2)
banana = Product("Banana", 3)

apple.price_with_tax()
```

Syntax Error

[파이썬 대표 예외]
`IndexError`
- list, tuple 존재하지 않는 인덱스

`ValueError`
- 문자열을 숫자로 변환 등 허용하지 않는 처리

`TypeError`
- 서로 다른 데이터 간 연산
- 호출할 수 없는 대상을 호출할 때

`ZeroDivisionError`

`try - except - else - finally`
- try: 예외 발생 가능
- except: 예외 발생했을 때 실행
- else: 로직 성공적일 때
- finally: 예외 발생, 로직 실패 상관없이 실행

`raise 예외명("메시지")`
- 직접 에러를 발생시켜 흐름 제어


Error Traceback
- from below to above 역순으로 읽기
- 맨 아래쪽 줄에 최종 예외의 이름 + 상세 원인 메시지

```python
def set_user_age(age):
    if age < 0:
        # 비즈니스 규칙에 맞지 않는 음수 입력 시 의도적으로 예외를 발생시킴
        raise ValueError("나이는 음수가 될 수 없습니다.")
    print(f"사용자의 나이가 {age}세로 설정되었습니다.")

try:
    input_age = -5
    set_user_age(input_age)
except ValueError as e:
    # 발생한 예외를 식별하여 메시지 출력
    print(f"[시스템 경고] {e}")
```

### 내장 함수
- `isinstance(varirable, type)`: is this variable the type? True/False

- `all(iterable)`: 데이터 내 모든 요소 True? = True 반환 | `all([])` = **True** (False가 하나도 없으므로 True)
- `any(iterable)`: 데이터 내 단 하나 True? = True 반환 ex) 여러 로그인 수단 중 하나만 연동되어 있어도 접근 허용 | `any([])` = **Fals** (True가 단 하나라도 존재하지 않으므로 False)

#### 통계 함수
- `len()`: 데이터 총 개수(길이) 반환, 문자열은 문자열 길이, 리스트/튜플은 요소 개수
- `sum(iterable)`: 요소의 합계(숫자로만 구성된)
- `round(self, num)`: 반올림, 소수점 뒤 num까지
- `math.ceil(..)`: 올림
- `math.floor(..)`: 버림

- `mean()`: 평균값
- `median()`: 중앙값

- `map(규칙(함수), iterable)`: iterable의 모든 요소에 규칙(함수)를 일괄 적용하여 새로운 형태로 반환 ex)일괄 수정, 전체 상품 할인율 적용 
- `filter(조건(함수), iterble))`: iterable의 모든 요소 중 조건(함수)에 맞는 요소만 걸러내어 추출

- `enumerate(iterable, start=n)`: iterable 순회하면서 인덱스 번호 자동 부여, 인덱스 번호는 start 옵션으로 지정하여 사용 가능
- `zip(iterableA, iterableB)`: 서로 다른 두 개의 iterables를 순서대로 짝 지어줌 (이름-이메일, 이름2-이메일2)

- `sorted(iterable)`: 기본값 오름차순으로 데이터 정렬 및 새로운 리스트 반환. 
  - `sorted(iterable, reverse=True)`: 내림차순 정렬
- `min(iterable)`: 최저값 요소 찾아냄
- `max(iterable)`: 최고값 요소 찾아냄

```python
# teacher's feedback on practice 08-1-3
class Hero:
    def __init__(self, name):
        self.name = name
        self.hp = 100

    def take_damage(self, damage):

        minus_damage = damage if self.hp >= damage else self.hp # 한줄 조건부 사용 가능. 3항 조건 구조. self.hp는 minus-damage로 대입됨
        self.hp -= minus_damage
        print("남은 체력: ", self.hp)

        if self.hp <= 0:
            print(f"{self.name} 캐릭터가 쓰러졌습니다")


cha = Hero("Spidy")
cha.take_damage(20)
```

```python
# teacher's feedback on practice 08-2-2
class DoorLock:
    def __init__(self, password):
        if not self.validate(password):
            return
        
        self.__password = str(password)

    def get_password(self):
        return self.__password

    def set_password(self, password):
        if not self.validate(password):
            return # 반환값 없더라도 return 하면 함수 종료, 다음 코드 실행시키지 않음
        
        self.__password = str(password)

    def validate(self, password):

        result = len(str(password)) >= 4
        if not result:
            print("경고: 비밀번호는 4자리 이상이어야 합니다")

        return result
```

```python
def order(order_num):
    if order_num <= 0:
        raise NoMinusZero
    print("주문 수량이 정상적으로 설정되었습니다.")
```
