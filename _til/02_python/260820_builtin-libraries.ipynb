json 데이터 표준 변환 - **s**에 주목

- `json.loads(s)` : deserialize (json str -> python obj)

*with open()시*

- `json.dump(obj, fp)` : serialize (python obj -> save as json file)
- `json.load(fp)` : deserialize (read jsoon file -> python obj)



json.dump() --> boolean 소문자 변환

```python
# dumps & loads

import json

Hero = [{
    "name": "Spiderman",
    "level": 50,
    "location": "Queens"
},
{
    "name": "Black Cat",
    "level": 50,
    "location": "Queens"
}]

# Serialize
serialize_withS = json.dumps(Hero, ensure_ascii=False)
print("json_book:", serialize_withS, "type:", type(serialize_withS))

deserialize = json.loads(serialize_withS)
```

```python
# dump & load with open()
import json

character_profile = {
    "name": "Spiderman",
    "level": 100,
    "authority": True
}

# serialize
with open("character_profile.json", "w", encoding="UTF-8") as fp:
    json.dump(character_profile, fp, indent=4) #########

# deserialize
with open("character_profile.json", "r", encoding="UTF-8") as fp:
    deserialized = json.load(fp) ##########

print("Deserialized:", deserialized, "| type:", type(deserialized))
```

#### with open()
- 메모리 자원 낭비 방지 
- close() 알아서 처리, 파일 자동으로 닫아줌

**파일 쓰기**

`with open("file", "w", encoding="UTF-8")`
- overwrite whatever you wrote before

`with open("file", "a", encoding="UTF-8")`
- append next to what you wrote before

**파일 읽기**

`with open("file", "r", encoding="UTF-8")`

- read(): 한꺼번에 전체 텍스트를 읽어 옴
- readline(): 한 줄 읽어오기, 한 줄씩 읽어서 메모리에 적재하므로 메모리 소비 적음
- realines(): 라인별로 읽어서 리스트로 반환, 각 라인은 하나의 요소, 한꺼번에 메모리에 올림 -> 메모리 소리 많음, 큰 파일에 적합하지 않음

```python
with open("test.txt", "w", encoding="UTF-8") as fp:  #"w" overwrites
    fp.write("TIL on")
    fp.write("with open()")

with open("test.txt", "a", encoding="UTF-8") as fp: # "a" appends
    fp.write("\npracitce 1")

with open("test.txt", "r", encoding="UTF-8") as fp: # easier to iterate
    for line in fp:
        print(line)
```

Comma Seperated Values (CSV) style
- 콤마로 데이터 구분
- 표 형태 구조 표현하는 텍스트 파일 형식 `행/열`
- 엑셀/데이터베이스 연동 쉬움
- 용량 작음
- 2차원 리스트 구조와 연결된 데이터 관리


write
- `csv.writer()`
- `writerow()`: one line
- `writerows()`: multiple lines

read
- `csv.reader()`
- list()를 통해 2차원 리스트로 변환



```python
import csv
import json

# CSV 파일 변환 실습
member_data = [
    ["id", "name", "score"],
    ["user1", "철수", "85"],
    ["user2", "영희", "90"]
]

# 파이썬 리스트 데이터를 CSV 파일로 내보냅니다. (직렬화)
with open("members.csv", "w", encoding="utf-8", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(member_data)

# 저장된 CSV 파일을 읽어와 데이터 구조를 확인합니다. (역직렬화)
with open("members.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    parsed_csv = list(reader)
    print("--- CSV 복원 결과 ---")
    print(parsed_csv)

# JSON 파일 변환 실습
course_data = {
    "course_name": "AX Agent Course",
    "active": True,
    "details": {"hours": 8, "location": "Incheon"}
}

# 파이썬 딕셔너리 데이터를 JSON 파일로 내보냅니다. (직렬화)
with open("course.json", "w", encoding="utf-8") as f:
    json.dump(course_data, f, indent=4)

# 저장된 JSON 파일을 읽어와 파이썬 객체로 복원합니다. (역직렬화)
with open("course.json", "r", encoding="utf-8") as f:
    parsed_json = json.load(f)
    print("\n--- JSON 복원 결과 ---")
    print(parsed_json)
    print(f"데이터 타입: {type(parsed_json)}")
```

```python
# chapter11-1 문제1

# 순차적 데이터 수집 sync code
import time

def download_page(collected_data, delay_time):
    print(f"{collected_data} 시작....")
    time.sleep(delay_time)
    print(f"....{collected_data} 완료!")

start_time = time.time()

download_page("메인 페이지 데이터", 2)
download_page("상세 페이지 데이터", 2)

end_time = time.time()
print(end_time - start_time)
```

```python
# chapter11-1 문제2
import time
import asyncio

async def download_page_async(data, delay_time):
    print(f"{data} 시작....")
    await asyncio.sleep(delay_time)
    print(f"....{data} 완료!")    


async def main():
    start_time = time.time()
    
    await asyncio.gather(
        download_page_async("메인 페이지 데이터", 2),
        download_page_async("상세 페이지 데이터", 2),
    )

    end_time = time.time()
    print("총 소요 시간:", end_time - start_time, "초")

await main()
```

```python
# chapter11-1 문제3
import asyncio
import time

async def fetch_weather(location, delay):
    print(f"{location}의 날씨 데이터 수집 시작...")
    await asyncio.sleep(delay) # await 으로 current loop 완료까지 freeze 되지 않고 next loop 으로 넘어감
    print(f"{location}의 날씨 데이터 수집 완료!")

async def main():
    start_time = time.time()

    await asyncio.gather(
        fetch_weather("서울", 1.5),
        fetch_weather("뉴욕", 2.5),
        fetch_weather("런던", 1.0),
    )

    end_time = time.time()
    print(f"총 소요 시간: {end_time - start_time:.1f}초") # time taken = the longest task
    

await main()
```
