## 2. Key Concepts

#### pandas
- `DataFrame.loc[]`: 행 **이름**으로 검색 
- `DataFrame.iloc[]`: **숫자**로 검색, '미만'

- 불리언 인덱싱: 조건 -> T/F 표기
- query()조건 검색: 조건에 해당하는 값만 필터링해서 보여줌


[결측치]
- `np.nan`, `pd,NA`, `NONE`
- `DataFrame.isna().sum()`: 컬럼별 결측치 총 개수, 기본값: axis=0(행 기준)
- `DataFrame.dropna()`: 결측치 삭제, 기본값 행 기준

[대치]
- `df_tofill.fillna(df_tobefilledinside)`
- `df.ffill()` 앞쪽 값으로 대체
- `df.bfill()` 뒤쪽 값으로 대체
- `df.interpolate(method='linear')`: 변화율 계산 -> 일정 간격에 따라 결측치 채움

[그룹화]
-`df.groupby(column)`
-`df.mean(numeric_only=True)`
-`df.to_numpy(dtype=numpy.자료형)`

### matplotlib `import matplotlib.pylot as plt`
- 캔버스: figure, 축:axes
- 산점도: 데이터 분포
- 꺾은선 그래프: 시계열 데이터의 흐름
- 히스토그램: 데이터 빈도수
- 박스플롯: 이상치 분포
- 캔버스 하나: `plt.feature(figsize=(캔버스 크기))`
- 캔버스 여러개: `plt.subplots(행, 열, figsize=(캔버스 크기))`
    - figsize(가로인치, 세로인치)

## 3. Practice 실습 코드

````python
# 1장 2강 문제 1
import pandas as pd

df = pd.read_csv("dataset.csv")

print("DataFrame 정보 출력:\n")
print(df.info()) 
print("=" * 50)
print("DataFrame 결측치 개수 출력(수정전):\n")
print(df.isnull().sum()) 

# 문자형 데이터 결측치 -> mode(최빈값)로 대체
mode_artists = df['artists'].mode()[0] 
df['artists'] = df['artists'].fillna(mode_artists)

mode_trackname = df['track_name'].mode()[0]
df['track_name'] = df['track_name'].fillna(mode_trackname)

# 숫자형 데이터 tempo, popularity 열 확인
df.loc[:, ['tempo', 'popularity']]


# tempo와 popularity의 0번째 데이터 수정
df.loc[0, ['tempo', 'popularity']] = 0

print("=" * 50)
print("DataFrame 최종 결측치 개수 출력(수정후):\n")
print(df.isnull().sum())ㄴ
````

````python
# 1장 2강 문제 1
import pandas as pd

df = pd.read_csv("dataset.csv") 

print("중복된 행 개수:", df.duplicated().sum())

df = df.drop_duplicates().reset_index(drop=True)  #변화된 데이터는 변수에 저장!

print("=" * 50)
print("필터링 전:")
print(df['duration_ms'].describe())

df = df.query('duration_ms >= 0 and duration_ms <= 1e6') #변화된 데이터는 변수에 저장!, e6= 0이 6개
print("=" * 50)
print("필터링 후:")
print(df['duration_ms'].describe())
````

````python
# 1장 2강 문제 3
import platform
from matplotlib import rc
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

os_name = platform.system()

if os_name == "Windows":
    rc("font", family="Malgun Gothic")
else:
     rc("font", family="Apple Gothic")

df = pd.read_csv("dataset.csv")
df = df[df['track_genre'].isin(['k-pop', 'techno'])] # track_genre 필터링

fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(18,5)) # 1행 3열의 그래프 영역, 영역 사이즈는 18x5, 서브플롯 3개

# histogram
ax1.hist(df['popularity'], bins=20, color='pink', edgecolor='black')
ax1.set_title("[histogram]")

# boxplot
sns.boxplot(x=df['track_genre'], y=df['danceability'], data=df, ax=ax2, color='skyblue')
            
# scatter
scatter = ax3.scatter(
    df['tempo'],
    df['popularity'],
    c=df['explicit'],
    cmap='RdYlGn',
    s=df['popularity'] / 2,
    alpha=0.4
)

ax3.set_xlabel("템포")
ax3.set_ylabel("인기도")

fig.colorbar(scatter, ax=ax3, label="명시적 가사 포함 여부")
````
