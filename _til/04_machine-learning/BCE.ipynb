# Binary Cross Entropy (BCE)

## 1. BCE란?

**Binary Cross Entropy(BCE)**는 **이진 분류(Binary Classification)**에서 사용하는 대표적인 **Loss Function**이다.

이진 분류에서는 정답이 두 가지이다.

```text
0 → Negative
1 → Positive
```

예시:

```text
스팸 메일인가?

No  → 0
Yes → 1
```

---

## 2. BCE의 역할

모델은 보통 클래스 자체가 아니라 **0~1 사이의 확률**을 예측한다.

```text
Actual (y)     → 실제 정답
Prediction (ŷ) → 모델이 예측한 확률
```

BCE는 다음을 측정한다.

> **실제 정답과 모델이 예측한 확률 사이의 차이, 즉 모델의 예측 오류를 측정한다.**

---

## 3. BCE 공식

$$
BCE = -\left[y\log(\hat{y}) + (1-y)\log(1-\hat{y})\right]
$$

| 기호 | 의미 |
|---|---|
| $y$ | 실제 정답 (`0` 또는 `1`) |
| $\hat{y}$ | 모델의 예측값 (`0~1`) |
| $\log$ | 로그 함수 |

---

## 4. 실제값이 1인 경우

실제 정답이:

$$
y = 1
$$

이라면 BCE 공식은:

$$
BCE = -\log(\hat{y})
$$

### 좋은 예측

```text
Actual = 1
Prediction = 0.99
```

실제값 `1`에 가깝게 예측했으므로:

> **Loss가 매우 작다.**

### 나쁜 예측

```text
Actual = 1
Prediction = 0.01
```

실제값은 `1`인데 `0`에 가깝게 예측했으므로:

> **Loss가 매우 크다.**

---

## 5. 실제값이 0인 경우

실제 정답이:

$$
y = 0
$$

이라면 BCE 공식은:

$$
BCE = -\log(1-\hat{y})
$$

### 좋은 예측

```text
Actual = 0
Prediction = 0.01
```

실제값 `0`에 가깝게 예측했으므로:

> **Loss가 작다.**

### 나쁜 예측

```text
Actual = 0
Prediction = 0.99
```

실제값은 `0`인데 `1`이라고 거의 확신했으므로:

> **Loss가 매우 크다.**

---

## 6. BCE의 핵심 특징

BCE의 핵심은 다음과 같다.

> **틀린 예측을 확신할수록 더 큰 Loss를 부여한다.**

예를 들어 실제값이 `1`일 때:

| Prediction | 의미 | Loss |
|---|---|---|
| `0.99` | 거의 정확하게 예측 | 매우 작음 |
| `0.8` | 비교적 정확 | 작음 |
| `0.5` | 확신 없음 | 중간 |
| `0.1` | 틀린 방향으로 예측 | 큼 |
| `0.001` | 완전히 틀린 방향으로 확신 | 매우 큼 |

```text
Actual = 1

Prediction
0 ───────────────────────────── 1
↑                               ↑
큰 Loss                      작은 Loss
```

# Sigmoid와 BCE

BCE는 일반적으로 **Sigmoid와 함께 사용된다.**

## 1. Linear Combination

먼저 모델은 입력 데이터 `X`를 이용하여 선형 계산을 수행한다.

$$
z = wX + b
$$

`z`는 다음처럼 어떤 값이든 될 수 있다.

```text
-10, -3, 0, 2, 10 ...
```

---

## 2. Sigmoid 적용

Sigmoid는 `z`를 **0~1 사이의 값으로 변환하는 Activation Function**이다.

$$
\hat{y} = \sigma(z)
$$

$$
\sigma(z) = \frac{1}{1+e^{-z}}
$$

예:

```text
z = -3 → Sigmoid → 0.047
z =  0 → Sigmoid → 0.5
z =  3 → Sigmoid → 0.953
```

즉, 모델의 출력값을 **확률처럼 해석할 수 있는 0~1 사이의 값**으로 변환한다.

---

## 3. 전체 학습 흐름

```text
Input X
   ↓
Linear Combination
z = wX + b
   ↓
Sigmoid Activation Function
   ↓
Prediction ŷ (0 ~ 1)
   ↓
Binary Cross Entropy (BCE)
   ↓
Calculate Loss
   ↓
Loss를 최소화하도록
Weight(w)와 Bias(b)를 업데이트
```

---

# 핵심 정리

- **Sigmoid**
  - Activation Function의 한 종류
  - 입력값을 `0~1` 사이의 값으로 변환
  - 이진 분류에서 예측 확률을 출력할 때 사용

- **BCE (Binary Cross Entropy)**
  - 이진 분류에서 사용하는 대표적인 Loss Function
  - 실제값(`y`)과 예측값(`ŷ`)의 차이를 측정
  - 잘못된 예측을 강하게 확신할수록 더 큰 Loss를 부여

## 한 줄 요약

> **Sigmoid는 모델의 출력값을 0~1 사이의 확률값으로 변환하고, BCE는 그 예측 확률이 실제 정답과 얼마나 다른지 측정하는 Loss Function이다.**
