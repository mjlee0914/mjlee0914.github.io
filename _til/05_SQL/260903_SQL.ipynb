## 2. Key Concepts

#### 데이터베이스 정규화
- 제1 정규화: "원자성" - 컬럼 하나에는 하나의 값만
- 제2 정규화: 복합 PK -> 각 PK를 컬럼별로 따로 의존하지 않도록 분리
- 제3 정규화: 일반 컬럼끼리 의존하지 않도록 분리

#### SQL 언어 범주
1. DDL Data Definition Language 데이터 정의어
   - CREATE table: 테이블 생성
   - ALTER table: 테이블 구조 변경
   - DROP table: 테이블 삭제
   - TRUNCATE table: 테이블 구조 유지 + 테이블 내용 삭제

2. DML Data Manipulation Language 데이터 조작어
    - INSERT: 행 추가
    - UPDATE
    - DELETE
    - DQL Data Query Language 데이터 질의어
      - SELECT: 데이터 view

3. DCL Data Control Language 데이터 제어
    - GRANT
    - REVOKE
  
4. TCL Transaction Control Language 
    - COMMIT: 데이터를 db에 영구 반영
    - ROLLBACK: commit 이전 시점 반영내용 되돌리기 (주문처리할 때)

#### Transaction의 특성
- 원자성: all or nothing
- 격리성
- 일관성
- 영속성

#### 제약조건 (선택)
1. PRIMARY KEY - 기본키, unique not null
2. NOT NULL
3. UNIQUE
4. DEFAULT ... - 값 안들어오면 디폴트값 자동 적용 ex) 현재날짜
5. REFERENCES - FK 참조 테이블 지정

#### 자동증감 serial
- numbering 자동화
- insert into에 굳이 넣지 않아도 자동 넘버링 적용

#### 복합키 composite key (mutliple PK)
- 여러 컬럼을 조합하여 PK 구성


- 복합키 적용 예시

- default 적용 예시

## 4. Daily Quest

[틀린 문제]
> FK가 참조하는 값이 반드시 참조 대상 테이블의 PK에 존재해야 한다는 규칙을 무엇이라고 하나요?

- 오답: 기본키 제약조건
- 정답: 참조 무결성
- 해설:
  - 개체 무결성(기본키 제약조건): PK는 unique + not null
  - 참조 무결성: 자식 테이블의 FK가 참조하는 값은 부모 테이블의 PK 값에 존재해야 한다. 
  - (FK, PK 각각 혼자 삭제될 수 없음. 
  - ON DELETE CASCADE를 사용하여 부모 테이블 PK값이 삭제되면 이를 참조하는 자식 테이블의 FK값도 마찬가지로 삭제되도록 설정
  - 혹은 ON DELETE SET NULL로 PK 값 삭제시 FK는 NULL값이 되도록 처리 가능)
