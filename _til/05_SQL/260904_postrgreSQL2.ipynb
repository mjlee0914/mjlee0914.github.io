## 2. Key Concepts

[참고]
- PostgreSQL 사용시 작은따옴표는 '일반 문자열' 전용, 큰따옴표의 경우 '특수문자가 포함된 컬럼이나 테이블명'에 사용
- PostgreSQL -> 대소문자 구분 X

#### DDL
**DDL은 삭제시 rollback이 되지 않음**
- CREATE
- ALTER
- DROP: 테이블 + 내용 전부 삭제
- TRUNCATE: 테이블 구조는 유지하고 내용만 삭제

[ALTER 종류]
- 컬럼 추가: `alter table 테이블명 add column 자료형 [제약조건];`
- 컬럼 삭제: `alter table 테이블명 drop column 컬럼명;`
- 컬럼 수정: `alter table 테이블명 rename column 기존컬럼명 to 변경컬럼명;`
- 컬럼 타입 변경: `alter table 테이블명 alter column 컬럼명 type 자료형;` **(데이터 없을 때 구조변경)**
- 컬럼 타입 변경: `alter table 테이블명 alter column 컬럼명 type 자료형 using (컬럼명::자료형);` **(데이터 있을 때 구조변경)**
- 컬럼 제약조건 변경: `alter table 테이블명 alter column 컬럼명 set 제약조건/default;`
- 컬럼 제약조건 삭제: `alter table 테이블명 alter column 컬럼명 drop 제약조건/default;`
- 컬럼 제약조건 추가: 
  - `alter table 테이블명 add constraint 제약조건구문별명 primary key(컬럼명, ...);`
  - `alter table 테이블명 add constraint 제약조건구문별명 foreign key(컬럼명, ...) references 참조할 테이블(컬럼명) on delete cascade | set null;`
  - `alter table 테이블명 add constraint 제약조건구문별명 unique(컬럼명, ...);`
  - `alter table 테이블명 add constraint 제약조건구문별명 primary key(컬럼명, ...);`
